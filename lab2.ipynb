   P(B | Test = Pozitiv) = P(Test = Pozitiv | B) * B / P(Test = Pozitiv) = 0,0095 / 0,1085 = 95/1085 = 19/217 ≈ 0,0876


c) P(B | Test = Pozitiv) >= 0,5 => P(Test = Pozitiv | B) * B / P(Test = Pozitiv) >= 0,5
   => 0,95*0,01 / ( 0,95*0,01 + (1-S)*0,99 ) >= 0,5 => 0,0095 >= (0,0095 + (1-S) * 0,99) * 0,5 
   => 0,0190 >= 0,0095 + 0,99 - 0,99 * S => 0,99*S >= 0,0095 + 0,99 - 0,0190 => 0,99*S >= 0,9805 => S >= 98,05/99 = 0,99(04)

```python
import numpy as np

#b)
P_B = 0.01
P_pos_given_B = 0.95
P_neg_given_not_B = 0.9

P_not_B = 1-P_B
P_pos_given_not_B = 1-P_neg_given_not_B

P_pos = P_pos_given_B*P_B + P_pos_given_not_B*P_not_B

P_B_given_pos = (P_pos_given_B*P_B) / P_pos

n=100000
bolnav = np.random.binomial(1, P_B, n)

test_pozitiv = np.where(
    bolnav == 1,
    np.random.binomial(1, P_pos_given_B, n),
    np.random.binomial(1, P_pos_given_not_B, n),
)

bolnavi_si_pozitivi = (bolnav == 1) & test_pozitiv

P_B_given_pos_simulat = np.sum(bolnavi_si_pozitivi) / np.sum(test_pozitiv)

print(f"Probabilitate teoretică P(B|+): {P_B_given_pos * 100:.2f}%")
print(f"Probabilitate simulată P(B|+):  {P_B_given_pos_simulat * 100:.2f}%")
```
